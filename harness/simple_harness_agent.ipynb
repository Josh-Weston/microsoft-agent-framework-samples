```python
import os
from azure.identity import DefaultAzureCredential
from agent_framework_openai import OpenAIChatClient
from agent_framework import create_harness_agent
from dotenv import load_dotenv

load_dotenv()
```

```python
base_url = os.getenv("FOUNDRY_OPENAI_BASE_URL")
if not base_url:
    raise ValueError("FOUNDRY_OPENAI_BASE_URL environment variable is not set.")

model = os.getenv("FOUNDRY_MODEL")
if not model:
    raise ValueError("FOUNDRY_MODEL environment variable is not set.")

openai_client = OpenAIChatClient(
    model=model,
    base_url=base_url,
    credential=DefaultAzureCredential(),
)
```

```python
agent = create_harness_agent(
    client=openai_client
)

session = agent.create_session()
response = await agent.run('Plan a weekend trip to Montreal', session=session)
print(response.text)
```

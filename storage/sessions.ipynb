## OpenAI / Foundry

Nothing is stored on the server for retrieval. Everything is returned as part of the conversation and needs to be repackaged as part of the conversation


```python
chat_completion_agent = utils.new_base_openai_completion_agent(
    name="MyAgent",
    description="A friendly and helpful assistant.",
    instructions="You are a friendly and helpful assistant."
)

chat_completion_session = chat_completion_agent.create_session()
service_session_id = chat_completion_session.service_session_id
chat_completion_session.to_dict()
```

```python
responses_agent = utils.new_base_openai_responses_agent(
    name="MyResponsesAgent",
    description="A friendly and helpful assistant for responses.",
    instructions="You are a friendly and helpful assistant for responses."
)
responses_session = responses_agent.create_session()
service_responses_session_id = responses_session.service_session_id
responses_session.to_dict()
```

From above, the framework creates a unique session_id for both, and their service_session_id is set to None.

**Chat Completions Does Not Receive a service_session_id**

```python
response = chat_completion_agent.run("My name is Josh", session=chat_completion_session, stream=True)
async for message in response:
    # print(message.to_dict())
    pass


print(chat_completion_session.to_dict()) # all messages are stored on "state" property of the session
final_response = await response.get_final_response()
print(final_response.to_dict())
```

**Responses Does Receive a service_session_id**

```python
response = responses_agent.run("My name is Josh", session=responses_session, stream=True)
async for message in response:
    # print(message.to_dict())
    pass


print(responses_session.to_dict()) # the "state" property is empty, messages are stored on the server
final_response = await response.get_final_response()
print(final_response.to_dict())
```

### Continue Chat Completions Session (full message history)

```python
chat_completion_agent = utils.new_base_openai_completion_agent(
    name="MyAgent",
    description="A friendly and helpful assistant.",
    instructions="You are a friendly and helpful assistant."
)

chat_completion_session = chat_completion_agent.create_session()
service_session_id = chat_completion_session.service_session_id
chat_completion_session.to_dict()
```

```python
response = chat_completion_agent.run("My name is Josh", session=chat_completion_session, stream=True)
async for message in response:
    # print(message.to_dict())
    pass


print(chat_completion_session.to_dict()) # all messages are stored on "state" property of the session
final_response = await response.get_final_response() # does not include the user's input message
print(final_response.to_dict())
```

```python
# Approach 1) Use AgentSession class
previous_session = chat_completion_session.to_dict()
new_session = AgentSession.from_dict(previous_session)
new_session.to_dict()
```

```python
# Approach 2) Use chat_completion_agent.create_session() and rebind
new_session = chat_completion_agent.create_session()
new_session = new_session.from_dict(previous_session) # from_dict returns a new session instance
new_session.to_dict()
```

```python
response = await chat_completion_agent.run("What is my name?", session=new_session)
for message in response.messages:
    print(message.to_dict())
```

### Responses No Session (stateless)

If you do not use a session, the ResponsesAPI works in the same was the ChatCompletionsAPI

```python
responses_agent = utils.new_base_openai_responses_agent(
    name="MyResponsesAgent",
    description="A friendly and helpful assistant for responses.",
    instructions="You are a friendly and helpful assistant for responses."
)
```

```python
user_message = Message(role="user", contents="My name is Josh")
response = await responses_agent.run("My name is Josh")
messages = [user_message] + [message for message in response.messages]
```

```python
messages = messages + [Message(role="user", contents="What is my name?")]
```

```python
response = await responses_agent.run(messages)
response.to_dict()
```

### Responses Session (in-memory stateless)

This is the same as above, except the framework handles the messages for you in its state, similar to how state works for ChatCompletionsAPI

```python
responses_agent = utils.new_base_openai_responses_agent(
    name="MyResponsesAgent",
    description="A friendly and helpful assistant for responses.",
    instructions="You are a friendly and helpful assistant for responses.",
    context_providers=[InMemoryHistoryProvider("memory", load_messages=True)]
)

session = responses_agent.create_session()
session.to_dict()
```

```python
response = await chat_completion_agent.run("My name is Josh", session=session)
response.to_dict()
```

```python
session.to_dict() # service_session_id is blank because we are using in-memory stateless
```

```python
response = await chat_completion_agent.run("What is my name?", session=session)
response.to_dict()
```
